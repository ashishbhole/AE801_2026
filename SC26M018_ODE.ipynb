
## Comprehensive Assignment on Numerical Solution to ODEs: Implementing the Leapfrog Method to Solve an IVP

### Overview
Every particle-in-cell (PIC) code advances its particles with the same core step: a leapfrog update of position and velocity under electric and magnetic fields. In this assignment, we write a leapfrog particle pusher from scratch and test it on the electron plasma oscillation problem, which sets the familiar PIC time-step limit. All quantities are in normalized units, so no physical constants are needed.

---

### 1. The Leapfrog Particle Pusher
A particle with charge-to-mass ratio $q/m$ obeys two coupled first-order ODEs:
$$\frac{dx}{dt} = v, \qquad \frac{dv}{dt} = \frac{q}{m}(E(x) + v \times B)$$

Leapfrog stores position at whole time levels $t_n = nh$ and velocity at half levels $t_{n+1/2} = \left(n + \frac{1}{2}\right)h$. Each step first updates the velocity using the force at the current position (**the kick**), then updates the position using the new velocity (**the drift**):
$$v_{n+1/2} = v_{n-1/2} + h \frac{q}{m} E_n$$
$$x_{n+1} = x_n + h v_{n+1/2}$$

The staggering makes the method second-order accurate with only one force evaluation per step. Because initial conditions give $v$ at $t = 0$, the velocity must first be moved back half a step to $v_{-1/2}$.

---

### 2. Electron Plasma Oscillation (Problem Formulation)
An electron displaced from equilibrium in a uniform, fixed ion background feels an electric field proportional to its displacement and oscillates at the plasma frequency $\omega_p$. In normalized units ($\omega_p = 1$), the governing equations and initial conditions are:

$$\frac{dx}{dt} = v, \qquad \frac{dv}{dt} = -\omega_p^2 x$$

* **Initial Conditions:** $x(0) = 0.5, \quad v(0) = 0$
* **Parameters & Domain:** $\omega_p = 1, \quad 0 \le t \le 20$
* **Exact Analytical Solution:**
  $$x(t) = 0.5 \cos(t), \qquad v(t) = -0.5 \sin(t)$$

### Mathematical Formulation and Discretization Logic

In a standard central-difference discretization around time level $t_n$, we usually take a step forward to $n+1$ and a step backward to $n-1$ over a total span of $2h$:
$$\left.\frac{dv}{dt}\right|_n \approx \frac{v_{n+1} - v_{n-1}}{2h}$$

In the **staggered leapfrog method**, we apply the exact same central-difference concept, except instead of stepping to $n+1$ and $n-1$, we step half a grid spacing forward ($n+1/2$) and half a grid spacing backward ($n-1/2$) around $t_n$. This shrinks the total span from $2h$ down to $h$:
$$\left.\frac{dv}{dt}\right|_n \approx \frac{v_{n+1/2} - v_{n-1/2}}{h}$$

Similarly, for the position derivative $\frac{dx}{dt} = v$, we center the difference around the half-step $t_{n+1/2}$ by stepping $+h/2$ forward to $n+1$ and $-h/2$ backward to $n$:
$$\left.\frac{dx}{dt}\right|_{n+1/2} \approx \frac{x_{n+1} - x_n}{h} = v_{n+1/2}$$

### Connection to the Electron Plasma Oscillation Equations

From the general Lorentz force equation given in the problem statement:
$$\frac{d\mathbf{v}}{dt} = \frac{q}{m}(\mathbf{E}(\mathbf{x}) + \mathbf{v} \times \mathbf{B}) = \frac{q}{m}\mathbf{F}$$
Applying our half-step central difference yields the kick-drift equations:
$$\mathbf{v}_{n+1/2} = \mathbf{v}_{n-1/2} + h \frac{q}{m} \mathbf{F}_n, \qquad \mathbf{x}_{n+1} = \mathbf{x}_n + h \mathbf{v}_{n+1/2}$$

For the 1D electron plasma oscillation in normalized units ($\omega_p = 1$, with no explicit physical constants $q$ or $m$ needed), the governing velocity ODE is:
$$\frac{dv}{dt} = -\omega_p^2 x$$

Comparing the right-hand sides, the force-per-unit-mass term $\frac{q}{m} F_n$ is simply the acceleration $a(x_n) = -\omega_p^2 x_n$. Substituting this into our staggered central-difference formulas gives the final update rules for the code:

1. **Half-Step Velocity Initialization ($t = -h/2$):**
   Since initial conditions are given at $t = 0$, we first step velocity backward by $h/2$ to start the staggered grid:
   $$v_{-1/2} = v_0 - \frac{h}{2} a(x_0) = v_0 + \frac{h}{2} \omega_p^2 x_0$$

2. **Kick Step (Velocity update centered at $t_n$):**
   $$v_{n+1/2} = v_{n-1/2} - h \omega_p^2 x_n$$

3. **Drift Step (Position update centered at $t_{n+1/2}$):**
   $$x_{n+1} = x_n + h v_{n+1/2}$$

```python
#defining the leapfrog function

def leapfrog(x,v,h,wp=1):
    v = v - h*wp**2*x
    x = x + h*v
    return x,v
```

```python
h = 0.1
t_end = 20
wp = 1
n = int(t_end/h)
t_x = np.linspace(0,t_end,n+1)
t_v = t_x[:-1]+0.5*h
x_0 = 0.5
v_0 = 0.0

a_0 = -(wp**2) * x_0
v = v_0 - 0.5 * h * a_0
x = x_0

x_num = np.zeros(n + 1)
v_num = np.zeros(n)
x_num[0] = x

for i in range(n):
    x,v = leapfrog(x,v,h,wp)
    v_num[i] = v
    x_num[i+1] = x

x_exact = 0.5 * np.cos(t_x)
v_exact = -0.5 * np.sin(t_v)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(16, 4.5))
axes[0].plot(t_x, x_exact, 'k-', label='Exact $x(t) = 0.5\cos(t)$')
axes[0].plot(t_x,x_num,'ro-',markersize=4, label ='Leapfrog $x_n$')
axes[0].set_xlabel('Time')
axes[0].set_ylabel('position')
axes[0].set_title('position vs time')
axes[0].legend()
axes[0].grid(True)

axes[1].plot(t_v, v_exact, 'k-', label='Exact $v(t) = -0.5\sin(t)$')
axes[1].plot(t_v,v_num,'ro-',markersize=4, label ='Leapfrog $v_n$')
axes[1].set_xlabel('Time')
axes[1].set_ylabel('velocity')
axes[1].set_title('velocity vs time')
axes[1].legend()
axes[1].grid(True)

axes[2].plot(x_exact,-0.5 * np.sin(t_x), 'k-', label='Exact $x(t) and v(t)$')
axes[2].plot(x_num[:-1], v_num,'ro-',markersize=4,label ='Leapfrog $x_n, v_n$')
axes[2].set_xlabel('position')
axes[2].set_ylabel('velocity')
axes[2].set_title('position vs velocity')
axes[2].legend()
axes[2].grid(True)

plt.tight_layout()
plt.show()
```

```python
h_values = [0.2, 0.1, 0.05, 0.025]

error_stag =[]

for h in h_values :

    t_end = 20
    wp = 1
    n = int(t_end/h)
    t_x = np.linspace(0,t_end,n+1)
    t_v = t_x[:-1]+0.5*h
    x_0 = 0.5
    v_0 = 0.0

    a_0 = -(wp**2) * x_0
    v = v_0 - 0.5 * h * a_0
    x = x_0

    x_num = np.zeros(n + 1)
    v_num = np.zeros(n)
    x_num[0] = x

    for i in range(n):
        x,v = leapfrog(x,v,h,wp)
        v_num[i] = v
        x_num[i+1] = x

    x_exact = 0.5 * np.cos(t_x)
    max_err = np.max(np.abs(x_num - x_exact))
    error_stag.append(max_err)

error_unstag = []

for h in h_values :

    t_end = 20
    wp = 1
    n = int(t_end/h)
    t_x = np.linspace(0,t_end,n+1)
    t_v = t_x[:-1]+0.5*h
    x_0 = 0.5
    v_0 = 0.0

    a_0 = -(wp**2) * x_0
    v = v_0
    x = x_0

    x_num = np.zeros(n + 1)
    v_num = np.zeros(n)
    x_num[0] = x

    for i in range(n):
        x,v = leapfrog(x,v,h,wp)
        v_num[i] = v
        x_num[i+1] = x

    x_exact = 0.5 * np.cos(t_x)
    max_err = np.max(np.abs(x_num - x_exact))
    error_unstag.append(max_err)

print("STAGGERED START")
print(f"{'h':<10} | {'Max Error e(h)':<20} | {'Order p'}")
print("-" * 45)
for i in range(len(h_values)):
    if i == 0:
        print(f"{h_values[i]:<10} | {error_stag[i]:<20.10f} | -")
    else:
        p = np.log2(error_stag[i-1] / error_stag[i])
        print(f"{h_values[i]:<10} | {error_stag[i]:<20.10f} | {p:.4f}")

print("\nUNSTAGGERED START")
print(f"{'h':<10} | {'Max Error e(h)':<20} | {'Order p'}")
print("-" * 45)
for i in range(len(h_values)):
    if i == 0:
        print(f"{h_values[i]:<10} | {error_unstag[i]:<20.10f} | -")
    else:
        p = np.log2(error_unstag[i-1] / error_unstag[i])
        print(f"{h_values[i]:<10} | {error_unstag[i]:<20.10f} | {p:.4f}")
```

**Analysis of the Observed Order of Accuracy**

* **Staggered Start:** When the half-step velocity is properly initialized using $v_{-1/2} = v_0 - \frac{h}{2} a(x_0)$, the maximum global error decreases by a factor of 4 each time $h$ is halved. This yields an observed convergence order of $p \approx 2$. This confirms that the staggered leapfrog method is a second-order accurate numerical scheme.
* **Unstaggered Start:** When the initialization is skipped and we set $v_{-1/2} = v_0$, a first-order $\mathcal{O}(h)$ truncation error is introduced during the very first step. Even though the leapfrog iteration itself is second-order accurate, this initial error propagates throughout the entire integration domain. As a result, the global observed order of accuracy degrades completely to $p \approx 1$.

### 5: Order of the Method (Central-Difference Equivalency)

To show that the staggered leapfrog scheme is equivalent to the central-difference formula, we start with the position update (drift) equation at step $n$:
$$x_{n+1} = x_n + h v_{n+1/2} \implies v_{n+1/2} = \frac{x_{n+1} - x_n}{h}$$

Writing the same position update equation for the previous step ($n-1$) gives:
$$x_n = x_{n-1} + h v_{n-1/2} \implies v_{n-1/2} = \frac{x_n - x_{n-1}}{h}$$

Now, substitute these two velocity expressions into the velocity update (kick) equation, $v_{n+1/2} = v_{n-1/2} - h \omega_p^2 x_n$:
$$\frac{x_{n+1} - x_n}{h} = \frac{x_n - x_{n-1}}{h} - h \omega_p^2 x_n$$

Multiply the entire equation by $h$:
$$x_{n+1} - x_n = x_n - x_{n-1} - h^2 \omega_p^2 x_n$$

Rearranging the terms yields the standard central-difference formula for $\frac{d^2x}{dt^2} = -\omega_p^2 x$:
$$x_{n+1} - 2x_n + x_{n-1} = -\omega_p^2 h^2 x_n$$

### : Stability Limit of the Leapfrog Method

**1. Derivation of the Stability Condition ($\omega_p h < 2$)**
From Task 5, the leapfrog scheme can be written as the difference equation:
$$x_{n+1} - (2 - \omega_p^2 h^2)x_n + x_{n-1} = 0$$

To find the stability criteria, we assume a trial solution of the form $x_n = \lambda^n$. Substituting this into the difference equation and dividing by $\lambda^{n-1}$ yields the characteristic quadratic equation:
$$\lambda^2 - (2 - \omega_p^2 h^2)\lambda + 1 = 0$$

Let $b = 1 - \frac{1}{2}\omega_p^2 h^2$. The equation becomes $\lambda^2 - 2b\lambda + 1 = 0$, which has roots:
$$\lambda = b \pm \sqrt{b^2 - 1}$$

For the numerical solution to be stable (producing bounded oscillations instead of exponential blow-up), we require the amplification factor magnitude to be bounded: $|\lambda| \le 1$. 
* If $|b| < 1$, the term under the square root is negative, yielding complex conjugate roots $\lambda = b \pm i\sqrt{1 - b^2}$. The magnitude is $|\lambda|^2 = b^2 + (1 - b^2) = 1$. The scheme is perfectly stable and conserves amplitude.
* If $|b| > 1$, one of the real roots will have a magnitude strictly greater than $1$, causing the solution to grow exponentially.

Therefore, absolute stability requires $|b| < 1$:
$$-1 < 1 - \frac{1}{2}\omega_p^2 h^2 < 1$$
$$-2 < -\frac{1}{2}\omega_p^2 h^2 < 0$$
$$0 < \omega_p^2 h^2 < 4 \implies \omega_p h < 2$$



```python
wph_values = [1.9, 1.99, 2.01, 2.1]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
axes = axes.flatten()

for idx, wph in enumerate(wph_values):
    
    wp = 1
    h = wph / wp
    n = 100
    
    x_0 = 0.5
    v_0 = 0.0
    
    a_0 = -(wp**2) * x_0
    v = v_0 - 0.5 * h * a_0
    x = x_0
    
    x_num = np.zeros(n + 1)
    v_num = np.zeros(n)
    x_num[0] = x
    
    for i in range(n):
        x,v = leapfrog(x,v,h,wp)
        v_num[i] = v
        x_num[i+1] = x
        
    axes[idx].plot(range(n + 1), x_num, 'ro-', markersize=4)
    axes[idx].set_title(f"$\omega_p h = {wph}$")
    axes[idx].set_xlabel('Step number (i)')
    axes[idx].set_ylabel('Position $x_i$')
    axes[idx].grid(True)

plt.tight_layout()
plt.show()
```
