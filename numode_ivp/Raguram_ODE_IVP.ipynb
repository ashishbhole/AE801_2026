# Approach and numerical method

#### Given condition:
The leapfrog particle pusher
A particle with charge-to-mass ratio q/m obeys two coupled first-order ODEs:
$$\frac{dx}{dt} =v,\frac{dv}{dt} = \frac{q}{m} (E(x)+v× B).\qquad\qquad(1)$$

Leapfrog stores position at whole time levels tn = nh and velocity at half levels $t_{n+\frac{1}{2}} = (n+\frac{1}{2})h.$
Each step first updates the velocity using the force at the current position (the kick), then
updates the position using the new velocity (the drift):

$$v_{n+\frac{1}{2}} = v_{n−\frac{1}{2}} +h\frac{q}{m}F_{n} , x_{n+1} = x_{n} +hv_{n+\frac{1}{2}}. \qquad\qquad(2)$$

The staggering makes the method second-order accurate with only one force evaluation per step.
Initial conditions give $v$ at $t = 0$, so the velocity must first be moved back half a step to $v_\frac{-1}{2}$.
Electron plasma oscillation
An electron displaced from equilibrium in a uniform, fixed ion background feels an electric field
proportional to its displacement, and oscillates at the plasma frequency ωp. In normalized units
($ω_p = 1$), the motion is

$$
\frac{dx}{dt} = v, \qquad
\frac{dv}{dt} = -\omega_p^2 x, \qquad
\omega_p = 1
$$

$$
x(0) = 0.5, \qquad
v(0) = 0, \qquad
0 \leq t \leq 20
$$

**Exact solution:**

$$
x(t) = 0.5\cos t, \qquad
v(t) = -0.5\sin t
$$

# Numerical algorithm

```python
"Leapfrog function"
def leapfrog(x_0,v_0,h):
    x = x_0
    v = v_0
    h = h
    a = -x
    
    v_n = v + (h*a)
    x_n = x + (h*v_n)

    
    return (x_n,v_n)
```

```python
"Staggered start"

h=np.array([0.025,0.05,0.1,0.2])
#print(h[0])
o=len(h)
X_max = []
V_max = []
X_rms = []
V_rms = []
for i in range(0,o):
    t = 20
    n = int(t/h[i])
    x = 0.5
    a = -x
    v_i = 0
    v = v_i - (0.5*h[i]*a)
    p = []
    p.append(v_i)
    q =[]
    q.append(x)

    t1 = np.arange(0,20+h[i],h[i])
    x_e = 0.5*np.cos(t1)
    v_e = -0.5*np.sin(t1)

    for j in range(0,n):
        s = leapfrog(x,v,h[i])
        p.append(s[1])
        q.append(s[0])
        v = s[1]
        x = s[0]

    v_diff = p-v_e  
    x_diff = q-x_e
    x_m    = np.max(np.abs(x_diff))
    x_r    = np.sqrt(np.mean(x_diff**2))
    v_m    = np.max(np.abs(v_diff))
    v_r    = np.sqrt(np.mean(v_diff**2))
    X_max.append(x_m)
    V_max.append(v_m)
    X_rms.append(x_r)
    V_rms.append(v_r)
    
#print(X_max,V_max)
    
#print(X_max)
power_staggered = []
for i in range(1,len(X_max)):
    ratio = (X_max[i]/X_max[i-1])
    power_staggered.append(int(np.log2(ratio)))
    
print(power_staggered)
```

## Finding the Order of the Method

 The equations given in the Leapfrog scheme are:

$$
v_{n+1/2} = v_{n-1/2} + h \cdot a(x_n) \quad \text{   (1)}
$$


$$
x_{n+1} = x_n + h \cdot v_{n + 1/2} \quad \text{   (2)}
$$

Rearranging equation (2) for velocity:

$$
v_{n + 1/2} = \frac{x_{n+1} - x_{n}}{h} \quad \text{   (3)}
$$

Now, we have $v_{n + 1/2}$ in terms of $x$,\
and for $v_{n - 1/2}$ we can use $n = 1 - 1/2 $ in equation (3) \
ie, moving backwards by one time step:

$$
v_{n - 1/2} = \frac{x_{n} - x_{n-1}}{h} \quad \text{   (4)}
$$

Substituting equations (3) and (4) into equation (1), and substituting the electron plasma acceleration $a(x_n) = -\omega_p^2 x_n$, we get:

$$
\frac{x_{n+1} - x_{n}}{h} = \frac{x_{n} - x_{n-1}}{h} - h\omega_p^2 x_n
$$

Multiplying the entire equation by $h$ to clear the denominators:

$$
x_{n+1} - x_{n} = x_{n} - x_{n-1} - \omega_p^2 h^2 x_n
$$


Gathering all position variables onto the left-hand side yields the final central-difference formula:

$$
\boxed{x_{n+1} - 2x_n + x_{n-1} = -\omega_p^2 h^2 x_n}
$$
