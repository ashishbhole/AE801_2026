```python

import numpy as np

# Given initial conditions
x0 = 0.5
v0 = 0.0

# Time step
h = 0.1

# Plasma frequency
omega_p = 1.0


# Function to calculate acceleration
def acceleration(x):
    return -(omega_p**2) * x


# Question 1: Leapfrog method
def leapfrog(x, v, h):
    a = acceleration(x)

    # Update velocity first
    v_new = v + h * a

    # Update position using the new velocity
    x_new = x + h * v_new

    return x_new, v_new


# Question 2: Calculate the initial half-step velocity
v_half = v0 - (h / 2) * acceleration(x0)

print("Initial half-step velocity:", v_half)


# Perform one Leapfrog step
x1, v_half = leapfrog(x0, v_half, h)

print("Position after one step:", x1)
print("Velocity at the half-step:", v_half)
```
