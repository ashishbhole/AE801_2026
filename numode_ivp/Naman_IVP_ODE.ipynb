### LeapFrog Integration Method 
LeapFrog Integration is a method for numerically integrating differential equations of the form: 

$$\ddot{x}= \frac{d^2 x}{dt^2} = a(x)$$
or equivalently, in the form
$$ 
\dot{v} = \frac{dv}{dt} = a(x),
\qquad
\dot{x} = \frac{dx}{dt} = v,
$$
particularly in the case of a dynamical system of classical mechanics.

In leapfrog integration, the equations for updating position and velocity are:
$$ a_{x} = f(x) $$
$$ v_{n + 1/2} = v_{n - 1/2} + ha_{x}  $$
$$ x_{n + 1} = x_{n} + hv_{n + 1/2 } $$
where $x_{n}$ is the current positon, $v_{n + 1/2}$ is the velocity at $n+1/2$ step and $h$ is the time step. 

In this assignemnt, we have been the position and velocity functions with time as: 
$$\dot{x} = v,
\qquad
\dot{v} = - \omega_{p}^2 x, 
\qquad
\omega_{p} = 1
$$
we have the initial conditions as: 
$$ 
x(0) = 0.5, 
\qquad
v(0) = 0, 
\qquad
0 \leq t \leq 20
$$

We need to compare this with the exact solution: 
$$
x(t) = 0.5\cos(t), 
\qquad
v(t) = -0.5 \sin(t)
$$

### Idea of Implementation
There are 6 parts to this:
1. Writing a leapfrog(x, v, h) function that performs one iteration and gives new x, v.
2.  Conmputing from the staggered start: $$v_{-1/2} = v_{0} - \frac{h}{2} a(x_{0}), \qquad a(x) = - \omega_{p}^2 x $$
3.  Integrate from $t \in [0, 20]$. Plot $x$ at whole levels and velocity at half levels. Plot the phase-space trajectory ($x$ against $v$).
4. Check for convergence for $h \in [0.2, 0.1, 0.05, 0.025]$. Tabulate the maximum error in x and estimate the observed order $ p = \log_{2}\frac{e(h)}{e(h/2)}$. Then repeat with the unstaggered start $v_{-1/2} = v_{0}$. Remark on what happens to the order and explain why.
5. Show LeapFrog in Central Difference Form.
6. Derive Stability Limit i.e. $\omega_{p}h < 2$. Confirm it by running with $\omega_{p}h \in [1.9, 1.99, 2.01, 2.1]$. 

```python
##Leap Frog Integration Method 
import numpy as np
import matplotlib.pyplot as plt

## PART - 1 
## Simple Leap Frog Integration for 1 iteration as asked in 1st requriement 
def leapfrog(x, v, h):
    v_half = v - 0.5 * h * x
    x_new = x + h * v_half
    v_new = v_half - h * x_new

    return x_new, v_new, v_half

x = np.array([0.5, 1.0, 1.5])
v = np.array([0.0, 0.2, -0.1])
h = 0.05

x_new, v_new, v_half = leapfrog(x, v, h)
print("New positions:", x_new)
print("New velocities:", v_new)
print("Half Time Step:", v_half)

##PART - 2, 3
##Comparision with Exact Solution
#Initial Conditions
omega_p = 1
h = 0.05
t = [0, 20]
n = int((t[1] - t[0]) / h)

timesteps_x = np.linspace(t[0], t[1], n+1)
timesteps_v = np.linspace(t[0], t[1], n)

x = np.zeros(n+1)
v_half = np.zeros(n)

x[0] = 0.5
v[0] = 0.0
v_minus_half = v[0] + 0.5*h*omega_p**2*x[0]

def leapfrog1(x, v, h):
    v_new = v - h * omega_p**2 * x
    x_new = x + h * v_new
    return x_new, v_new

for i in range(n):
    v_half[i] = v_minus_half
    x[i+1], v_minus_half = leapfrog1(x[i], v_minus_half, h)

# Exact solutions at corresponding times
x_exact = 0.5 * np.cos(omega_p * timesteps_x)
v_exact = -0.5 * np.sin(omega_p * timesteps_v)

# Plot position at whole time levels
plt.figure(figsize=(6, 3))
plt.plot(timesteps_x, x, label="Leapfrog")
plt.plot(timesteps_x, x_exact, "--", label="Exact solution")
plt.xlabel("Time (s)")
plt.ylabel("Position x")
plt.title("Position vs Time")
plt.legend()
plt.grid(True)
plt.show()

# Plot velocity at half time levels
plt.figure(figsize=(6, 3))
plt.plot(timesteps_v, v_half, label="Leapfrog")
plt.plot(timesteps_v, v_exact, "--", label="Exact solution")
plt.xlabel("Time (s)")
plt.ylabel("Velocity v")
plt.title("Half-Step Velocity vs Time")
plt.legend()
plt.grid(True)
plt.show()

## Phase Space Trajectory 
plt.figure(figsize=(6, 6))
plt.plot(x[0:n], v_half, label="Leapfrog")
plt.plot(x_exact[0:n], v_exact, "--", label="Exact solution")
plt.xlabel("Position x")
plt.ylabel("Velocity v")
plt.title("Phase Space Trajectory")
plt.legend()
plt.grid(True)
plt.show()
```

```python
## PART - 4

## Convergence Analysis 
#Staggered Start
t = [0, 20]
h_values = [0.2, 0.1, 0.05, 0.025]
x2 = {}
time = {}
x_exact_arr = {}
err_arr_1 = {}

for h in h_values:
    n_step = int((t[1] - t[0])/ h)
    timestep = np.linspace(t[0], t[1], n_step + 1)

    x_exact_1 = 0.5 * np.cos(omega_p * timestep)
    x1 = np.zeros(n_step + 1)
    v_half1 = np.zeros(n_step)

    x1[0] = 0.5
    v_0 = 0
    v_minus_half1 = v_0 + 0.5 * h * omega_p**2 * x1[0]

    for j in range(n_step):
        v_half1[j] = v_minus_half1
        x1[j+1], v_minus_half1 = leapfrog1(x1[j], v_minus_half1, h)

    x2[h] = x1
    time[h] = timestep
    x_exact_arr[h] = x_exact_1

    err_arr = np.zeros(n_step + 1)
    err_arr = (x1 - x_exact_1)
    err_arr_1[h] = err_arr
    
fig, axs = plt.subplots(2, 2, figsize=(8, 5), sharex=True, sharey=True)
for i in range(len(h_values)):
    ax = axs.flat[i]
    h = h_values[i]
    ax.plot(time[h], x_exact_arr[h], "k--", linewidth=1.2, label="Exact")
    ax.plot(time[h], x2[h], label="Leapfrog")
    ax.set_title(f"h = {h}")
    ax.set_xlabel("Time (s)")
    ax.set_ylabel("Position x")
    ax.legend()
    ax.grid(True)

fig.suptitle("Leapfrog vs Exact Solution for Different h (Staggered)", fontsize=14)
plt.tight_layout()
plt.show()

fig, axs = plt.subplots(2, 2, figsize=(8, 5), sharex=True, sharey=True)
for i in range(len(h_values)):
    ax = axs.flat[i]
    h = h_values[i]
    ax.plot(time[h], err_arr_1[h], "k--", linewidth=1.2, label="Error")
    ax.set_title(f"h = {h}")
    ax.set_xlabel("Time (s)")
    ax.set_ylabel("Error")
    ax.legend()
    ax.grid(True)

fig.suptitle("Errors for Different h (Staggered)", fontsize=14)
plt.tight_layout()
plt.show()

#Unstaggered Start 
x21 = {}
time1 = {}
x_exact_arr1 = {}
err_arr_2 = {}

for h in h_values:
    n_step = int((t[1] - t[0])/ h)
    timestep = np.linspace(t[0], t[1], n_step + 1)

    x_exact_1 = 0.5 * np.cos(omega_p * timestep)
    x1 = np.zeros(n_step + 1)
    v_half1 = np.zeros(n_step)

    x1[0] = 0.5
    v_0 = 0
    v_minus_half1 = v_0

    for j in range(n_step):
        v_half1[j] = v_minus_half1
        x1[j+1], v_minus_half1 = leapfrog1(x1[j], v_minus_half1, h)

    x21[h] = x1
    time1[h] = timestep
    x_exact_arr1[h] = x_exact_1

    err_arr1 = np.zeros(n_step + 1)
    err_arr1 = (x1 - x_exact_1)
    err_arr_2[h] = err_arr1
    
fig, axs = plt.subplots(2, 2, figsize=(8, 5), sharex=True, sharey=True)
for i in range(len(h_values)):
    ax = axs.flat[i]
    h = h_values[i]
    ax.plot(time[h], x_exact_arr1[h], "k--", linewidth=1.2, label="Exact")
    ax.plot(time[h], x21[h], label="Leapfrog")
    ax.set_title(f"h = {h}")
    ax.set_xlabel("Time (s)")
    ax.set_ylabel("Position x")
    ax.legend()
    ax.grid(True)

fig.suptitle("Leapfrog vs Exact Solution for Different h (Unstaggered)", fontsize=14)
plt.tight_layout()
plt.show()

fig, axs = plt.subplots(2, 2, figsize=(8, 5), sharex=True, sharey=True)
for i in range(len(h_values)):
    ax = axs.flat[i]
    h = h_values[i]
    ax.plot(time1[h], err_arr_2[h], "k--", linewidth=1.2, label="Error")
    ax.set_title(f"h = {h}")
    ax.set_xlabel("Time (s)")
    ax.set_ylabel("Error")
    ax.legend()
    ax.grid(True)

fig.suptitle("Errors for Different h (Unstaggered)", fontsize=14)
plt.tight_layout()
plt.show()

##Error Ratio for Staggered and Unstaggered Start
e1 = []
e2 = []
for h in h_values:
    e1.append(np.max(np.abs(err_arr_1[h])))
    e2.append(np.max(np.abs(err_arr_2[h])))

e1 = np.array(e1)
e2 = np.array(e2)

plt.figure(figsize=(6, 3))
plt.plot(h_values, e1,'-o', label="Staggered")
plt.plot(h_values, e2, '-o',label="Unstaggered")
plt.xlabel("Time Step h (s)")
plt.ylabel("Error in x")
plt.title("Error for different h")
plt.gca().invert_xaxis()
plt.legend()
plt.grid(True)
plt.show()

print("Staggered p:  ", np.log2(e1[:-1] / e1[1:]))
print("Unstaggered p:", np.log2(e2[:-1] / e2[1:]))
```

We see that as we move from staggered to unstaggered, the order of accuracy decreases from **second-order** to **first-order**. 
To understand, we use the Taylor Series:
$$ v(t) = v(0) + t \frac{dv}{dt}\bigg|_{t=0} + \frac{t^2}{2!} \frac{d^2v}{dt^2}\bigg|_{t=0} + O(t^3) $$
For Staggered Part, the system equation for velodcity is : 
$$ v_{-1/2} = v_{0} + h\frac{dv}{dt} $$
By comparing it with Taylor Series, we can it is as:
$$ v_{n+1/2} = v_{0} + h\frac{dv}{dt} + O(t^2) $$
This means that it is a second order accurate equation, the truncating error is of the order $t^2$. 

But for the Unstaggered Part, the equation is:
$$ v_{-1/2} = v_{0} $$
In the form of Taylor Series, this can be written as: 
$$ v_{n+1/2} = v_{0} + O(h) $$
This is a first order accurate equation, the truncating error is of the order $t$. 


### PART - 5: Velocity $v$ can be eliminated to show the scheme in central difference formula 

We have the 2 uncoupled equation from the LeapFrog Integration Method:

$$
\begin{aligned}
\mathbf{v}_{n+1/2} &= \mathbf{v}_{n-1/2} + ha(x), \\
\mathbf{x}_{n+1} &= \mathbf{x}_n + h\mathbf{v}_{n+1/2}.
\end{aligned}
$$

where 
$
a(x) = -\omega_p^2 \mathbf{x}_{n}
$

Velocity can be written in form of position and time step $h$ as:
$$
\mathbf{v}_{n+1/2} = \frac{x_{n+1} - x_{n}}{h}
$$ 
and 

$$
\mathbf{v}_{n-1/2} = \frac{x_{n} - x_{n-1}}{h}
$$ 

Put the velocity as a function of postion in orignal Leapfrog Equation. This will change the equation as:
$$
\frac{x_{n+1} - x_{n}}{h} = \frac{x_{n} - x_{n-1}}{h} - h\omega_p^2 \mathbf{x}_{n}
$$

Rearranging this:
$$
\frac{x_{n+1} - x_{n}}{h} - \frac{x_{n} - x_{n-1}}{h} = - h\omega_p^2 \mathbf{x}_{n}
$$

$$
\frac{x_{n+1} - x_{n} - x_{n} + x_{n-1}}{h} = - h\omega_p^2 \mathbf{x}_{n}
$$

$$
x_{n+1} - 2x_{n}  + x_{n-1} = - h^2\omega_p^2 \mathbf{x}_{n}
$$

This is the central difference form of the LeapFrog Integration. 

### PART - 6: Stability Limit 
We assume a solution of the central difference equation of the form: $x_{n} = r^{n}$
The central difference form becomes: 

$$
r^{n+1} + (h^2\omega_p^2 - 2)r^{n}  + r^{n-1} = 0
$$

Dividing by $r^{n-1}$, this will give us: 

$$
r^{2} + (h^2\omega_p^2 - 2)r + 1 = 0
$$ 

The roots can be found by solving this quadratic, we get:

$$
r_{1,2} = \frac{2 - h^2\omega_p^2 \pm \sqrt{(h^2\omega_p^2 - 2)^2 - 4}}{2}
$$

For stability, $\sqrt{(h^2\omega_p^2 - 2)^2 - 4} < 0$:
we get:

$$
(h^2\omega_p^2 - 2)^2 < 4
$$

$$
-2 < (h^2\omega_p^2 - 2) < 2
$$

$$
0 < h^2\omega_p^2 < 4
$$

$$
0 < h\omega_p < 2
$$

By taking the right hand limit, we can get the stability limit:
$$
h\omega_p < 2
$$

```python
#Staggered Start
t = [0, 20]
omega_p = 1
h_values = [1.9, 1.99, 2.01, 2.1]
x2 = {}
time = {}
x_exact_arr = {}
err_arr_1 = {}

for h in h_values:
    n_step = int((t[1] - t[0])/ h)
    timestep = np.linspace(t[0], t[1], n_step + 1)

    x_exact_1 = 0.5 * np.cos(omega_p * timestep)
    x1 = np.zeros(n_step + 1)
    v_half1 = np.zeros(n_step)

    x1[0] = 0.5
    v_0 = 0
    v_minus_half1 = v_0 + 0.5 * h * omega_p**2 * x1[0]

    for j in range(n_step):
        v_half1[j] = v_minus_half1
        x1[j+1], v_minus_half1 = leapfrog1(x1[j], v_minus_half1, h)

    x2[h] = x1
    time[h] = timestep
    x_exact_arr[h] = x_exact_1

    err_arr = np.zeros(n_step + 1)
    err_arr = (x1 - x_exact_1)
    err_arr_1[h] = err_arr
    
fig, axs = plt.subplots(2, 2, figsize=(8, 5), sharex=True, sharey=True)
for i in range(len(h_values)):
    ax = axs.flat[i]
    h = h_values[i]
    ax.plot(time[h], x_exact_arr[h], "k--", linewidth=1.2, label="Exact")
    ax.plot(time[h], x2[h], label="Leapfrog")
    ax.set_title(f'$\omega_p h = {h}$')
    ax.set_xlabel("Time (s)")
    ax.set_ylabel("Position x")
    ax.legend()
    ax.grid(True)

fig.suptitle("Stability Check for different $\omega_{p}h$ (Staggered)", fontsize=14)
plt.tight_layout()
plt.show()
```

This stability check is important because this gives a limit for a stable system. If this limit is exceeded then, the system goes unstable as shown in above plots. The upper limit was $\omega_{p}h < 2$ and as it is exceeded, the system goes unstable, it diverges more and more and within this limit, the system is stable no matter how close it is to the limit.
