```python
import numpy as np
def G(Y):
  return np.array ([Y[1],Y[2],-0.5*Y[0]*Y[2]])

def RK3(Y, h):

    K1 = G(Y)

    K2 = G(Y + h*K1/2)

    K3 = G(Y - h*K1 + 2*h*K2)

    Y_new = Y + h*(K1 + 4*K2 + K3)/6

    return Y_new

def solve(s, h):

    eta_inf = 10
    N = int(eta_inf/h)

    # Initial conditions
    Y = np.array([0.0, 0.0, s])

    eta = np.linspace(0, eta_inf, N+1)
    Y_values = np.zeros((N+1, 3))

    Y_values[0] = Y

    for i in range(N):
        Y = RK3(Y, h)
        Y_values[i+1] = Y

    return eta, Y_values

def residual (s,h):
  eta_inf , Y_values = solve(s,h)
  r = Y_values[-1][1] - 1
  return r                ## if d>0 f'(10)>1

def bisection(s_low, s_high, h, tol):

    r_low = residual(s_low, h)
    r_high = residual(s_high, h)

    if r_low * r_high > 0:
        print("Invalid initial interval")
        return None

    while abs(s_high - s_low) > tol:

        s_mid = (s_low + s_high)/2

        r_mid = residual(s_mid, h)

        if r_low * r_mid < 0:
            s_high = s_mid
            r_high = r_mid

        else:
            s_low = s_mid
            r_low = r_mid

    return (s_low + s_high)/2


# Initial guesses
s_low = 0.2
s_high = 0.4

h = 0.1
tol = 1e-6

s_correct = bisection(s_low, s_high, h, tol)

print("Correct value of s =", s_correct)
```
