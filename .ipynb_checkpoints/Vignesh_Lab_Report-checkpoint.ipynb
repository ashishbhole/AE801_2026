## 1. Aim

To generate a NACA four-digit airfoil and reconstruct its upper and lower surfaces using Lagrange interpolation and natural cubic spline interpolation. The two methods are compared using plots and interpolation errors.

## 2. Introduction

Airfoil shape is important in aerodynamic analysis. In this experiment, a NACA four-digit airfoil is generated using its standard geometric equations. A set of 12 chordwise sample points is then used to estimate the upper and lower surfaces. Lagrange interpolation and natural cubic spline interpolation are applied, and their results are compared with the original profile.

## 3. Theory

### 3.1 NACA four-digit airfoil

For a NACA four-digit airfoil, the first digit represents the maximum camber as a percentage of chord, the second digit gives the location of maximum camber in tenths of chord, and the last two digits represent the maximum thickness as a percentage of chord.

The thickness distribution is

\[
y_t=5t(0.2969\sqrt{x}-0.1260x-0.3516x^2+0.2843x^3-0.1015x^4).
\]

The mean camber line is

\[
y_c=\begin{cases}\dfrac{m}{p^2}(2px-x^2), & x\leq p,\\[4pt]
\dfrac{m}{(1-p)^2}[(1-2p)+2px-x^2], & x>p.
\end{cases}
\]

The upper and lower surface coordinates are calculated from the camber line, thickness distribution, and camber-line slope. The chord length is normalized to one.

### 3.2 Lagrange interpolation

Lagrange interpolation constructs a polynomial through the given data points:

\[
P(x)=\sum_{i=0}^{n-1} y_i L_i(x), \qquad
L_i(x)=\prod_{j=0,\,j\ne i}^{n-1}\frac{x-x_j}{x_i-x_j}.
\]

The polynomial passes through all sample points. However, using one polynomial over the whole interval can produce unwanted oscillations.

### 3.3 Natural cubic spline interpolation

A cubic spline uses a separate cubic polynomial between each pair of adjacent sample points. The pieces are joined so that the function and its first and second derivatives are continuous. For a natural cubic spline, the second derivatives at both endpoints are zero.

### 3.4 Error calculation

The maximum absolute error and root-mean-square (RMS) error are used to compare the methods:

\[
E_{max}=\max_i |y_i-\hat y_i|, \qquad
E_{RMS}=\sqrt{\frac{1}{N}\sum_{i=1}^{N}(y_i-\hat y_i)^2}.
\]

The errors are calculated for the upper and lower surfaces and combined for an overall comparison.

## 4. Procedure

1. Define the NACA four-digit airfoil equations.
2. Generate the original airfoil profile.
3. Select 12 chordwise sample points.
4. Apply Lagrange interpolation to the upper and lower surfaces.
5. Apply natural cubic spline interpolation to the same sample points.
6. Plot both reconstructed profiles with the original airfoil.
7. Calculate and compare the maximum and RMS errors.

## 5. Python implementation

The following cells contain the Python code used to generate the airfoil, perform both interpolation methods, plot the results, and calculate the errors.


```python
import numpy as np
import matplotlib.pyplot as plt

def naca4(m, p, t, x):
    """Calculate upper and lower surface coordinates of a NACA four-digit airfoil."""
    x = np.asarray(x, dtype=float)

    yt = 5 * t * (
        0.2969 * np.sqrt(x)
        - 0.1260 * x
        - 0.3516 * x**2
        + 0.2843 * x**3
        - 0.1015 * x**4
    )

    yc = np.where(
        x <= p,
        m / p**2 * (2 * p * x - x**2),
        m / (1 - p)**2 * ((1 - 2 * p) + 2 * p * x - x**2)
    )

    dyc = np.where(
        x <= p,
        2 * m / p**2 * (p - x),
        2 * m / (1 - p)**2 * (p - x)
    )

    theta = np.arctan(dyc)
    xu = x - yt * np.sin(theta)
    yu = yc + yt * np.cos(theta)
    xl = x + yt * np.sin(theta)
    yl = yc - yt * np.cos(theta)

    return xu, yu, xl, yl


# Generate a reproducible set of NACA four-digit parameters.
rng = np.random.default_rng(24)
camber_digit = int(rng.integers(1, 10))
location_digit = int(rng.integers(1, 10))
thickness_digit = int(rng.integers(6, 19))

m = camber_digit / 100
p = location_digit / 10
t = thickness_digit / 100
naca_name = f"{camber_digit}{location_digit}{thickness_digit:02d}"

print(f"Airfoil selected: NACA {naca_name}")
print(f"Maximum camber m = {m:.2f}")
print(f"Camber location p = {p:.1f}")
print(f"Maximum thickness t = {t:.2f}")

# Generate the reference profile.
x = np.linspace(0, 1, 400)
xu, yu, xl, yl = naca4(m, p, t, x)

plt.figure(figsize=(9, 4))
plt.plot(xu, yu, label="Upper surface")
plt.plot(xl, yl, label="Lower surface")
plt.xlabel("x/c")
plt.ylabel("y/c")
plt.title(f"Original NACA {naca_name} Airfoil")
plt.axis("equal")
plt.grid(True)
plt.legend()
plt.show()
```

## 6. Lagrange interpolation

The upper and lower surface ordinates are interpolated separately using the selected sample points. The reconstructed profile is plotted against the original airfoil.


```python
def Lagrng(x_data, y_data, n, z):
    """Evaluate the Lagrange interpolation polynomial at z."""
    total = 0.0

    for i in range(n):
        product = y_data[i]
        for j in range(n):
            if i != j:
                product *= (z - x_data[j]) / (x_data[i] - x_data[j])
        total += product

    return total


# Select sample points from the chord.
n_sample = 12
x_sample = np.linspace(0, 1, n_sample)
_, yu_s, _, yl_s = naca4(m, p, t, x_sample)

x_fine = np.linspace(x_sample[0], x_sample[-1], 400)
n = len(x_sample)

yu_lagrange = np.array([Lagrng(x_sample, yu_s, n, z) for z in x_fine])
yl_lagrange = np.array([Lagrng(x_sample, yl_s, n, z) for z in x_fine])

# Reference ordinates evaluated on the same chordwise grid.
yu_reference = np.interp(x_fine, x, yu)
yl_reference = np.interp(x_fine, x, yl)

plt.figure(figsize=(10, 5))
plt.plot(xu, yu, "k-", label="Original upper surface")
plt.plot(xl, yl, "k-", label="Original lower surface")
plt.plot(x_fine, yu_lagrange, "r--", label="Lagrange upper")
plt.plot(x_fine, yl_lagrange, "b--", label="Lagrange lower")
plt.scatter(x_sample, yu_s, s=28, label="Upper sample points")
plt.scatter(x_sample, yl_s, s=28, label="Lower sample points")
plt.xlabel("x/c")
plt.ylabel("y/c")
plt.title(f"NACA {naca_name}: Original and Lagrange Reconstruction")
plt.axis("equal")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()


def calculate_errors(y_reference, y_approx):
    error = y_reference - y_approx
    maximum = np.max(np.abs(error))
    rms = np.sqrt(np.mean(error**2))
    return error, maximum, rms


error_upper_lag, max_upper_lag, rms_upper_lag = calculate_errors(
    yu_reference, yu_lagrange
)
error_lower_lag, max_lower_lag, rms_lower_lag = calculate_errors(
    yl_reference, yl_lagrange
)

combined_lagrange_errors = np.concatenate(
    (error_upper_lag, error_lower_lag)
)
max_combined_lag = np.max(np.abs(combined_lagrange_errors))
rms_combined_lag = np.sqrt(np.mean(combined_lagrange_errors**2))

print(f"NACA Airfoil: {naca_name}")
print("\nLAGRANGE INTERPOLATION ERRORS")
print(f"Upper surface maximum error: {max_upper_lag:.2e}")
print(f"Upper surface RMS error:     {rms_upper_lag:.2e}")
print(f"Lower surface maximum error: {max_lower_lag:.2e}")
print(f"Lower surface RMS error:     {rms_lower_lag:.2e}")
print(f"Combined maximum error:      {max_combined_lag:.2e}")
print(f"Combined RMS error:          {rms_combined_lag:.2e}")
```

## 7. Natural cubic spline interpolation

The natural cubic spline is calculated for the same sample points. The endpoint second derivatives are set to zero, and the reconstructed profile is plotted against the original airfoil.


```python
def calculate_spline_derivatives(x_data, y_data):
    """Calculate second derivatives for a natural cubic spline."""
    n = len(x_data)

    if n < 3:
        raise ValueError("At least 3 sample points are required.")
    if len(y_data) != n:
        raise ValueError("x_data and y_data must have equal lengths.")
    if np.any(np.diff(x_data) <= 0):
        raise ValueError("x_data must be strictly increasing.")

    h = np.diff(x_data)
    A = np.zeros((n - 2, n - 2))
    b = np.zeros(n - 2)

    for i in range(1, n - 1):
        row = i - 1
        b[row] = 6 * (
            (y_data[i + 1] - y_data[i]) / h[i]
            - (y_data[i] - y_data[i - 1]) / h[i - 1]
        )
        A[row, row] = 2 * (h[i - 1] + h[i])

        if i > 1:
            A[row, row - 1] = h[i - 1]
        if i < n - 2:
            A[row, row + 1] = h[i]

    M_inner = np.linalg.solve(A, b)

    # Natural boundary conditions: second derivatives at endpoints are zero.
    M = np.zeros(n)
    M[1:-1] = M_inner
    return h, M


def cubic_spline_value(x_data, y_data, h, M, xq):
    """Evaluate a natural cubic spline at a single query point."""
    if xq < x_data[0] or xq > x_data[-1]:
        raise ValueError("Query point is outside the sample range.")

    i = np.searchsorted(x_data, xq, side="right") - 1
    i = max(0, min(i, len(x_data) - 2))

    term1 = M[i] * (x_data[i + 1] - xq)**3 / (6 * h[i])
    term2 = M[i + 1] * (xq - x_data[i])**3 / (6 * h[i])
    term3 = (
        (y_data[i] - M[i] * h[i]**2 / 6)
        * (x_data[i + 1] - xq) / h[i]
    )
    term4 = (
        (y_data[i + 1] - M[i + 1] * h[i]**2 / 6)
        * (xq - x_data[i]) / h[i]
    )

    return term1 + term2 + term3 + term4


h_upper, M_upper = calculate_spline_derivatives(x_sample, yu_s)
h_lower, M_lower = calculate_spline_derivatives(x_sample, yl_s)

yu_spline = np.array([
    cubic_spline_value(x_sample, yu_s, h_upper, M_upper, z)
    for z in x_fine
])
yl_spline = np.array([
    cubic_spline_value(x_sample, yl_s, h_lower, M_lower, z)
    for z in x_fine
])

plt.figure(figsize=(10, 5))
plt.plot(xu, yu, "k-", label="Original upper surface")
plt.plot(xl, yl, "k-", label="Original lower surface")
plt.plot(x_fine, yu_spline, "r--", label="Cubic spline upper")
plt.plot(x_fine, yl_spline, "b--", label="Cubic spline lower")
plt.scatter(x_sample, yu_s, s=28, label="Upper sample points")
plt.scatter(x_sample, yl_s, s=28, label="Lower sample points")
plt.xlabel("x/c")
plt.ylabel("y/c")
plt.title(f"NACA {naca_name}: Original and Cubic Spline Reconstruction")
plt.axis("equal")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()


error_upper_spline, max_upper_spline, rms_upper_spline = calculate_errors(
    yu_reference, yu_spline
)
error_lower_spline, max_lower_spline, rms_lower_spline = calculate_errors(
    yl_reference, yl_spline
)

combined_spline_errors = np.concatenate(
    (error_upper_spline, error_lower_spline)
)
max_combined_spline = np.max(np.abs(combined_spline_errors))
rms_combined_spline = np.sqrt(np.mean(combined_spline_errors**2))

print(f"NACA Airfoil: {naca_name}")
print("\nNATURAL CUBIC SPLINE ERRORS")
print(f"Upper surface maximum error: {max_upper_spline:.2e}")
print(f"Upper surface RMS error:     {rms_upper_spline:.2e}")
print(f"Lower surface maximum error: {max_lower_spline:.2e}")
print(f"Lower surface RMS error:     {rms_lower_spline:.2e}")
print(f"Combined maximum error:      {max_combined_spline:.2e}")
print(f"Combined RMS error:          {rms_combined_spline:.2e}")
```

## 8. Comparison of errors

The maximum absolute error and RMS error are calculated on the same evaluation grid for both methods. The values are displayed in scientific notation with two significant digits so that small errors remain visible.


```python
methods = ["Lagrange", "Natural cubic spline"]
max_errors = [max_combined_lag, max_combined_spline]
rms_errors = [rms_combined_lag, rms_combined_spline]

print(f"{'Method':24s} {'Maximum error':>16s} {'RMS error':>16s}")
for method, max_err, rms_err in zip(methods, max_errors, rms_errors):
    print(f"{method:24s} {max_err:16.2e} {rms_err:16.2e}")

plt.figure(figsize=(8, 4))
plt.bar(methods, rms_errors)
plt.ylabel("Combined RMS error")
plt.title("Interpolation error comparison")
plt.grid(axis="y")
plt.tight_layout()
plt.show()
```

## 9. Results and discussion

The plots show how closely each method reconstructs the original airfoil from the 12 sample points. The maximum error gives the largest difference at any evaluated point, while the RMS error describes the overall difference across the profile. The error table can be used to identify which method gives the closer reconstruction for this case.

Lagrange interpolation uses one polynomial across the full interval. A natural cubic spline uses separate cubic polynomials and joins them smoothly. The comparison depends on the selected airfoil and sample points. The natural boundary condition also sets the second derivative to zero at both endpoints, which may not exactly represent the airfoil curvature there.
